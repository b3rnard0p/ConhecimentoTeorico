```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

# 1. Importar & Preparar
df = pd.read_csv('/content/drive/MyDrive/Ciência de Dados/Dados/Arquivos framingham_heart_disease.csv')
df = df.dropna()

# Separar variáveis preditoras (X) e a variável alvo (y)
X = df.drop(columns=['TenYearCHD'])
y = df['TenYearCHD']

# 2. Treinar & Testar
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

# 3. Ajustar (Fit)
# modelo = LogisticRegression(max_iter=1000)
modelo = LogisticRegression(max_iter=2000)
modelo.fit(X_train, y_train)

# 4. Prever & Extrair Probabilidades
y_pred = modelo.predict(X_test)
y_proba = modelo.predict_proba(X_test)[:, 1]

# 5. Avaliação
acuracia = accuracy_score(y_test, y_pred)
matriz_confusao = confusion_matrix(y_test, y_pred)

print(f"Acurácia Global: {acuracia:.4f}")
print("Matriz de Confusão:\n", matriz_confusao)
```

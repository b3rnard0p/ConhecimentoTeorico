```python
import pandas as pd

caminho = '/content/drive/MyDrive/Ciência de Dados/Notas.xlsx'
arquivo = pd.read_excel(caminho)

# arquivo.info()
# arquivo.describe()
print(arquivo)
```

```python
arquivo = arquivo.fillna(0)

arquivo['Nota2'] = arquivo ['Nota2'].astype(float)
arquivo['Nota3'] = arquivo ['Nota3'].astype(float)
arquivo.info()
```

```python
# arquivo.info()
# arquivo.describe()
arquivo.iloc[0]
```

```python
print(len(arquivo))
tamanho = len(arquivo)

for indice in range(tamanho):
  print(arquivo.iloc[indice])
```

```python
#Média de cada nota

media_notas = arquivo[['Nota1', 'Nota2', 'Nota3']].mean()

print("Médias das Notas:")
for nota, media in media_notas.items():
    print(f"{nota}: {media:.2f}")
```

```python
#Média de cada aluno

arquivo['Media_Notas'] = arquivo[['Nota1', 'Nota2', 'Nota3']].mean(axis=1)
nota_minima_aprovacao = 6.0
arquivo['Situação'] = arquivo['Media_Notas'].apply(lambda x: 'aprovado' if x >= nota_minima_aprovacao else 'reprovado')
display(arquivo)
```

```python
caminho_salvar = '/content/drive/MyDrive/Ciência de Dados/Notas_Atualizadas.xlsx'
arquivo.to_excel(caminho_salvar, index=False)
print(f"DataFrame salvo com sucesso em: {caminho_salvar}")
```

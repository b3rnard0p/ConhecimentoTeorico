```python
import pandas as pd

caminho = '/content/drive/MyDrive/Ciência de Dados/Censo.xlsx'

arquivo = pd.read_excel(caminho)

#print(arquivo.dtypes)
print(arquivo.loc[arquivo['Cidade'] == 'Florianópolis', 'Salário Médio'])
```

```python
arquivo.drop('Estado', axis=1, inplace=True) #Deleta Coluna
print(arquivo)
```

```python
# arquivo.df.colums = [''] //Mudar nome das colunas
```

```python
arquivo.insert(2, 'Estado', ['PR', 'SP', 'SC', 'PR', 'SP']) #Cria coluna podendo escolher local

print(arquivo)
```

```python
print(arquivo.iloc[:, [0, 2]])
```

```python
estados_frequentes = arquivo['Estado'].mode() #Percebe quem repeti

for estado in estados_frequentes:
    quantidade = (arquivo['Estado'] == estado).sum() #Conta quanto tem de cada repetido
    print(f"Estado: {estado}, Quantidade: {quantidade}")
```

```python
cidades_sp = arquivo.query("Estado == 'SP'") #Acha as cidades do esatdo filtrado
print(cidades_sp)
```

```python
cidades_salario_alto = arquivo.query("`Salário Médio` > 1700")[['Cidade', 'Salário Médio']]
display(cidades_salario_alto)
```

```python
media_salarial_por_estado = arquivo.groupby('Estado')['Salário Médio'].mean() #Forma grupos de cidades de mesmo estado e calcula a média entre elas
print(media_salarial_por_estado)
```

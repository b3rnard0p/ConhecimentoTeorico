```python
import pandas as pd

caminho = '/content/drive/MyDrive/Ciência de Dados/Dados/Cidades.xlsx'
arquivo = pd.read_excel(caminho)
```

```python
arquivo = arquivo.set_index('Cidade')
porto_alegre_data = arquivo.loc['Porto Alegre']
display(porto_alegre_data)

idh_sp_salvador = arquivo.loc[['São Paulo', 'Salvador'], 'IDH']
display(idh_sp_salvador)
```

```python
primeira_linha = arquivo.iloc[0]
display(primeira_linha)
```

```python
duas_primeiras_linhas_colunas = arquivo.iloc[0:1, 0:1]
display(duas_primeiras_linhas_colunas)
```

```python
idh_belo_horizonte = arquivo.at['Belo Horizonte', 'IDH']
display(f"valor: {idh_belo_horizonte}")
```

```python
valor = arquivo.iat[-1, -1]
display(f"Valor: {valor}")
```

```python
cidades_filtradas = arquivo.loc[arquivo['Habitantes'] > 2000000, ['UF', 'IDH']].reset_index(drop=True)
display(cidades_filtradas)
```

```python
habitantes_por_uf = arquivo.groupby('UF')['Habitantes'].sum()
display(habitantes_por_uf)
```

```python
if 'Cidade' in arquivo.index.names:
    arquivo = arquivo.reset_index()

new_cities_df = pd.DataFrame([
    {'Cidade': 'Guaruja do Sul', 'UF': 'SC', 'Habitantes': 5000, 'IDH': 0.700},
    {'Cidade': 'Caçapava do Sul', 'UF': 'RS', 'Habitantes': 30000, 'IDH': 0.720},
    {'Cidade': 'São Sepe', 'UF': 'RS', 'Habitantes': 25000, 'IDH': 0.500}
])

arquivo = pd.concat([arquivo, new_cities_df], ignore_index=True)

arquivo['Categoria'] = arquivo['Habitantes'].apply(lambda x: 'Metrópole' if x > 1000000 else 'Interior')
display(arquivo)
```

```python
idh_por_categoria = arquivo.groupby('Categoria')['IDH'].agg(['min', 'mean', 'max'])
display(idh_por_categoria)
```

```python
contagem_distribuida = arquivo.groupby(['UF', 'Categoria']).size()
display(contagem_distribuida)
```

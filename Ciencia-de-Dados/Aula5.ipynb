```python
import pandas as pd
import seaborn as sns
import plotly.express as px

caminho = '/content/drive/MyDrive/Ciência de Dados/Dados/censo2022.xlsx'
arquivo = pd.read_excel(caminho)
```

```python
arquivo_modificado = arquivo.iloc[3:]
display(arquivo_modificado.head(5))
display(arquivo.tail(6).head(5))
```

```python
regioes = arquivo_modificado.head(5)
display(regioes)
```

```python
#Peguei a coluna com o nome das regiões
regiao = regioes.columns[0]

#Pedi a coluna de regiões, e a coluna de superior completo de homem e de mulher
grafico = regioes[[regiao, 'Unnamed: 8', 'Unnamed: 12']].copy()

#Renomeando para legibilidade
grafico.columns = ['Região', 'Homens', 'Mulheres']

display(grafico)

#Estetica
fig = px.bar(
    long,
    x='Região',
    y='Percentual (%)',
    color='Sexo',
    barmode='group',
    title='Comparação: Superior Completo entre Homens e Mulheres por Região (2022)',
    text_auto='.1f',
    color_discrete_map={'Homens': '#1f77b4', 'Mulheres': '#d62728'}
)

fig.update_layout(xaxis_title="Região")

#Exibindo
fig.show()
```

```python
import matplotlib.pyplot as plt

regiao = regioes.columns[0]
grafico = regioes[[regiao, 'Unnamed: 4', 'Unnamed: 8', 'Unnamed: 12']].copy()

grafico.columns = ['Região', 'Total', 'Homens', 'Mulheres']

grafico.set_index('Região', inplace=True)

grafico.plot(
    kind='pie',
    subplots=True,
    figsize=(18, 6),
    autopct='%1.1f%%',
    legend=False,
    title=['Total', 'Homens', 'Mulheres']
)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

niveis_instrucao = [
    'Sem instrução/Fund. incomp.',
    'Fund. comp./Médio incomp.',
    'Médio comp./Sup. incomp.',
    'Superior completo'
]

linha_alvo = regioes.iloc[1]
nome_regiao = linha_alvo.iloc[0]

dados_pizza = pd.DataFrame({
    'Total': linha_alvo.iloc[1:5].astype(float).values,
    'Homens': linha_alvo.iloc[5:9].astype(float).values,
    'Mulheres': linha_alvo.iloc[9:13].astype(float).values
}, index=niveis_instrucao)

dados_pizza.plot(
    kind='pie',
    subplots=True,
    figsize=(18, 6),
    autopct='%1.1f%%',
    legend=False,
    title=[f'Total - {nome_regiao}', f'Homens - {nome_regiao}', f'Mulheres - {nome_regiao}']
)

plt.legend(niveis_instrucao, title="Nível de Instrução", bbox_to_anchor=(1.2, 0.5), loc="center left")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df_estados = arquivo_modificado.copy()

col_nomes = df_estados.columns[0]
df_estados[col_nomes] = df_estados[col_nomes].astype(str).str.strip()

estado_alvo = 'Rio de Janeiro'

niveis_instrucao = [
    'Sem instrução/Fund. incomp.',
    'Fund. comp./Médio incomp.',
    'Médio comp./Sup. incomp.',
    'Superior completo'
]

filtro_estado = df_estados[df_estados[col_nomes] == estado_alvo]

if filtro_estado.empty:
    print(f"Não encontrei '{estado_alvo}'. Verifique se o nome da variável original é mesmo 'arquivo_modificado'.")
else:
    linha_estado = filtro_estado.iloc[0]

    dados_pizza = pd.DataFrame({
        'Total': linha_estado.iloc[1:5].astype(float).values,
        'Homens': linha_estado.iloc[5:9].astype(float).values,
        'Mulheres': linha_estado.iloc[9:13].astype(float).values
    }, index=niveis_instrucao)

    dados_pizza.plot(
        kind='pie',
        subplots=True,
        figsize=(18, 6),
        autopct='%1.1f%%',
        legend=False,
        title=[f'Total - {estado_alvo}', f'Homens - {estado_alvo}', f'Mulheres - {estado_alvo}']
    )

    plt.legend(niveis_instrucao, title="Nível de Instrução", bbox_to_anchor=(1.2, 0.5), loc="center left")

    plt.tight_layout()
    plt.show()
```
